```python
from gliner import GLiNER

# Carga el modelo (la primera vez descargará los pesos desde Hugging Face)
model = GLiNER.from_pretrained("gliner-community/gliner_small-v2.5", load_tokenizer=True)

# Texto de prueba y etiquetas que quieres extraer
text = "Joan va anar a Barcelona amb el seu gos el mes passat."
labels = ["persona", "lloc", "animal"]

# Predicción de entidades
entities = model.predict_entities(text, labels)

for entity in entities:
    print(f"{entity['text']} => {entity['label']}")
```
